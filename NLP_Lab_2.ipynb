```python
text="Thiss PHONE is Amazng!!! I buyed it last wek, and the battry life is sooo goood but the camra quality isnt that greaat."
```

Text Normalization

```python
from nltk.tokenize import word_tokenize

tokens = word_tokenize(text)
print(tokens)
```

```python
text= re.sub(r"[^\w\s]", "", text)
print(text)
```

```python
lower_tokens = [word.lower() for word in tokens]
print(lower_tokens)
```

Tokenization

```python
from nltk.tokenize import word_tokenize

tokens = word_tokenize(text)
print(tokens)
```

```python
nlp = spacy.load('en_core_web_sm')
doc = nlp(text)
for token in doc:
    print(token.text)
```

Removing Stop words

```python
nltk.download('stopwords')
from nltk.corpus import stopwords

stop_words = set(stopwords.words('english'))
filtered_tokens = [word for word in lower_tokens if word.isalpha() and word not in stop_words]
print(filtered_tokens)
```

```python
nlp = spacy.load('en_core_web_sm')
doc = nlp(text)
for token in doc:
    print(token.text, '→',token.is_stop)
```

Stemming using NLTK Porter Stemmer

```python
from nltk.stem import PorterStemmer

ps = PorterStemmer()
stemmed_words = [ps.stem(word) for word in filtered_tokens]
print(stemmed_words)
```

Lemmatization using NLTK WordNet Lemmatizer and spaCy lemmatizer

```python
nltk.download('wordnet')
from nltk.stem import WordNetLemmatizer

lemmatizer = WordNetLemmatizer()
lemmatized_words = [lemmatizer.lemmatize(word) for word in filtered_tokens]
print(lemmatized_words)
```

```python
nlp = spacy.load('en_core_web_sm')
doc = nlp(text)
for token in doc:
    print(token.text, '→', token.lemma_)
```

Compute edit distance using NLTK with substitution cost = 1 and = 2

```python
s1 = "buyed"
s2 = "buy"
dist_nltk = nltk.edit_distance(s1, s2)
print(f"--- Standard NLTK (Sub=1) ---")
print(f"Distance ('{s1}' -> '{s2}'): {dist_nltk}")
```

```python
dist = nltk.edit_distance("buyed", "buy", substitution_cost=2)
print(f"Edit Distance: {dist}")
```
