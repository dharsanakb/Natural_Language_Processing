```python
import spacy

# Load model
nlp = spacy.load("en_core_web_sm")

# Input text
text = """The new phone has a fantastic camera but the battery life is terrible.
The screen is bright, but the price is high."""

doc = nlp(text)

print("Feature Pairs Detected:")

for token in doc:
    # Check if token is adjective and next token is noun
    if token.pos_ == "ADJ":
        for child in token.head.children:
            if child.pos_ == "NOUN":
                print(f"- ({token.text}, {child.text})")

    # Alternative: Direct ADJ → NOUN sequence
    if token.pos_ == "ADJ" and token.nbor(1).pos_ == "NOUN":
        print(f"- ({token.text}, {token.nbor(1).text})")
```

```python
def redact_sensitive_info(text):
    doc = nlp(text)
    redacted_text = text

    for ent in doc.ents:
        if ent.label_ in ["PERSON", "GPE"]:
            redacted_text = redacted_text.replace(ent.text, "[REDACTED]")

    return redacted_text

# Input text
text = """Startups in Bangalore are growing fast. Mr. Rajesh Kumar met with
Sarah Jones in London to discuss the merger."""

# Output
print(redact_sensitive_info(text))
```

```python
from sklearn.metrics import precision_score, recall_score, f1_score

# True tags
y_true = ["O", "DRUG", "O", "DRUG", "O", "O"]

# Model predictions
y_pred = ["O", "O", "O", "DRUG", "O", "DRUG"]

# Convert to binary (DRUG = 1, O = 0)
y_true_binary = [1 if tag == "DRUG" else 0 for tag in y_true]
y_pred_binary = [1 if tag == "DRUG" else 0 for tag in y_pred]

# Calculate metrics
precision = precision_score(y_true_binary, y_pred_binary)
recall = recall_score(y_true_binary, y_pred_binary)
f1 = f1_score(y_true_binary, y_pred_binary)

print("Precision:", precision)
print("Recall:", recall)
print("F1 Score:", f1)
```
